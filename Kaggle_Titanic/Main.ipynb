Viewing Dataset

```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

train_data = pd.read_csv("data/train.csv")
train_data.head()
```

```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)

test_data = pd.read_csv("data/test.csv")
test_data.head()
```

AI Model

```python
# Import the RandomForestClassifier from the sklearn.ensemble module
from sklearn.ensemble import RandomForestClassifier 

# Extract the target variable 'Survived' from the training data
y = train_data["Survived"]

# Define the feature columns to be used for training
features = ["Pclass", "Sex", "SibSp", "Parch"]

# Convert categorical variables into dummy/indicator variables (one-hot encoding) for the training data
X = pd.get_dummies(train_data[features])

# Similarly, convert categorical variables into dummy/indicator variables for the test data
X_test = pd.get_dummies(test_data[features])
```

```python
# Initialize the RandomForestClassifier model with 100 trees (n_estimators), 
# a maximum depth of 5 for each tree, and a random state for reproducibility
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)

# Train the model on the training features (X) and the target variable (y)
model.fit(X, y)

# Use the trained model to make predictions on the test data (X_test)
predictions = model.predict(X_test)
```

Predictions

```python
# Create a DataFrame to store the PassengerId from the test data and the predictions (Survived)
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})

# Export the results to a CSV file
output.to_csv('submission.csv', index=False)
```
